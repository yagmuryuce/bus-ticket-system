```python
admin_bilgileri = {"admin": "1234"}

kullanicilar = {
    "yağmur": "1111",
    "ayse": "2222"
}

seferler = {
    1: {"nereden": "Fethiye", "nereye": "İzmir", "saat": "09:00", "fiyat": 500, "koltuk": 5},
    2: {"nereden": "Fethiye", "nereye": "Ankara", "saat": "14:00", "fiyat": 800, "koltuk": 3},
    3: {"nereden": "Fethiye", "nereye": "İstanbul", "saat": "22:00", "fiyat": 1200, "koltuk": 2}
}

biletler = []


def seferleri_goster():
    print("\n--- SEFERLER ---")
    for no, s in seferler.items():
        print(f"""
Sefer No: {no}
Nereden : {s["nereden"]}
Nereye  : {s["nereye"]}
Saat    : {s["saat"]}
Fiyat   : {s["fiyat"]} TL
Koltuk  : {s["koltuk"]}
""")


def dosyaya_kaydet(bilet):
    with open("biletler.txt", "a", encoding="utf-8") as f:
        f.write(str(bilet) + "\n")


# ---------------- ÖDEME SİSTEMİ (GÜNCELLENDİ) ----------------

def odeme_yap():
    while True:
        print("\n1- Nakit\n2- Kredi Kartı")
        secim = input("Seçim: ")

        if secim == "1":
            return "Nakit"

        elif secim == "2":

            kart_no = input("Kart numarası: ")
            skt = input("SKT (MMYY): ")
            cvv = input("CVV: ")
            sifre = input("Kart şifre: ")

            # sadece sayı kontrolü
            if not kart_no.isdigit():
                print("Kart numarası sadece sayı olmalı")
                continue

            if not skt.isdigit():
                print("SKT sadece sayı olmalı")
                continue

            if not cvv.isdigit():
                print("CVV sadece sayı olmalı")
                continue

            if not sifre.isdigit():
                print("Şifre sadece sayı olmalı")
                continue

            # uzunluk kontrolleri
            if len(kart_no) < 12:
                print("Kart numarası çok kısa")
                continue

            if len(skt) != 4:
                print("SKT 4 hane olmalı (MMYY)")
                continue

            if len(cvv) != 3:
                print("CVV 3 hane olmalı")
                continue

            print("Kart ödeme başarılı")
            return "Kredi Kartı"

        else:
            print("Hatalı seçim")


# ---------------- BİLET AL ----------------

def bilet_al(kullanici):
    seferleri_goster()

    try:
        secim = int(input("Sefer no: "))
    except:
        print("Sayı gir")
        return

    if secim not in seferler:
        print("Geçersiz sefer")
        return

    if seferler[secim]["koltuk"] <= 0:
        print("Koltuk yok")
        return

    print("\nÖdeme ekranı")
    odeme = odeme_yap()

    seferler[secim]["koltuk"] -= 1

    bilet = {
        "kullanici": kullanici,
        "sefer": secim,
        "nereden": seferler[secim]["nereden"],
        "nereye": seferler[secim]["nereye"],
        "saat": seferler[secim]["saat"],
        "fiyat": seferler[secim]["fiyat"],
        "odeme": odeme
    }

    biletler.append(bilet)
    dosyaya_kaydet(bilet)

    print("\n--- BİLET OLUŞTU ---")
    print(f"""
Kullanıcı : {kullanici}
Nereden   : {bilet["nereden"]}
Nereye    : {bilet["nereye"]}
Saat      : {bilet["saat"]}
Fiyat     : {bilet["fiyat"]}
Ödeme     : {bilet["odeme"]}
""")


    while True:
        print("\n1- Yeni bilet\n2- Çıkış")
        c = input("Seçim: ")

        if c == "1":
            bilet_al(kullanici)
            return
        elif c == "2":
            return
        else:
            print("Hatalı")


def user_panel(kullanici):
    while True:
        print("\n1- Seferler\n2- Bilet al\n3- Çıkış")
        s = input("Seçim: ")

        if s == "1":
            seferleri_goster()
        elif s == "2":
            bilet_al(kullanici)
        elif s == "3":
            return
        else:
            print("Hatalı")


def admin_panel():
    while True:
        print("\n1- Seferler\n2- Biletler\n3- Çıkış")
        s = input("Seçim: ")

        if s == "1":
            seferleri_goster()
        elif s == "2":
            print(biletler)
        elif s == "3":
            return
        else:
            print("Hatalı")


while True:
    print("\n1- User\n2- Admin\n3- Çıkış")
    secim = input("Seçim: ")

    if secim == "1":
        u = input("Kullanıcı: ")
        p = input("Şifre: ")

        if u in kullanicilar and kullanicilar[u] == p:
            user_panel(u)
        else:
            print("Hatalı")

    elif secim == "2":
        u = input("Admin: ")
        p = input("Şifre: ")

        if u in admin_bilgileri and admin_bilgileri[u] == p:
            admin_panel()
        else:
            print("Hatalı admin")

    elif secim == "3":
        break

    else:
        print("Hatalı seçim")
```
